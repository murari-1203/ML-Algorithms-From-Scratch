```python
import numpy as np
```

```python
class LinearRegression:
    def __init__(self, learning_rate = 0.01, max_itr = 1000):
        self.bias = None
        self.weights = None
        self.learning_rate = learning_rate
        self.max_itr = max_itr

    def fit(self, X, y):
        m, n = X.shape # m - > samples, n -> featuers
        
        self.bias = 0
        self.weights = np.zeros(n)

        for i in range(self.max_itr):
            y_pred = self.bias + np.dot(X, self.weights)

            # Gradients
            derivitaive_with_bias = (1 / m) * np.sum(y_pred - y)
            derivative_with_weights = (1 / m) * np.dot(X.T, (y_pred - y))

            # Convergence Theorem
            self.bias = self.bias - self.learning_rate * derivitaive_with_bias
            self.weights = self.weights - self.learning_rate * derivative_with_weights

    def predict(self, X):
        y_pred = np.dot(X, self.weights)
        return y_pred
    
         
```

```python
# testing with Values X And y (y = 3 * X)
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([3, 6, 9, 12, 15])

model = LinearRegression();
model.fit(X, y)

y_pred = model.predict(np.array([[8], [10]]))
print(y_pred)
```
